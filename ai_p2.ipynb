```python
map_rooms = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["D"],
    "D": ["G"],
    "G": []
}

def bfs(start, goal):
    to_do = [start]
    visited = []
    order = []
    while to_do:
        room = to_do.pop(0)
        if room in visited:
            continue
        visited.append(room)
        order.append(room)
        if room == goal:
            return order
        for nxt in map_rooms[room]:
            if nxt not in visited and nxt not in to_do:
                to_do.append(nxt)
    return order

def dfs(start, goal):
    to_do = [start]
    visited = []
    order = []
    while to_do:
        room = to_do.pop() # pop from end for DFS
        if room in visited:
            continue
        visited.append(room)
        order.append(room)
        if room == goal:
            return order
        for nxt in map_rooms[room]:
            if nxt not in visited and nxt not in to_do:
                to_do.append(nxt)
    return order

# Define the cost of opening each door
door_costs = {
    ("A", "B"): 1, # Assuming '8' was a typo for 'B' as per the map
    ("B", "D"): 1,
    ("A", "C"): 5, # this door is heavy and costs 5
    ("C", "D"): 1,
    ("D", "G"): 1, # Assuming '6' was a typo for 'G' as per the map
}

# A helper to add up the cost of a whole path
def path_cost(path):
    total = 0
    for i in range(len(path) - 1):
        door = (path[i], path[i+1])
        if door not in door_costs:
            print(f"Warning: Door {door} not found in door_costs. This path is invalid.")
            return float('inf') # Return infinity for invalid paths
        total += door_costs[door]
    return total

path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]

print("Calculating path costs:")
print(f"Path 1 A -> B -> D -> G costs: {path_cost(path1)}")
print(f"Path 2 A -> C -> D -> G costs: {path_cost(path2)}")

cost1 = path_cost(path1)
cost2 = path_cost(path2)

# Determine the cheaper path
if cost1 < cost2:
    best_path = path1
    best_cost = cost1
else:
    best_path = path2
    best_cost = cost2

print("\nPath 1 cost:", cost1)
print("Path 2 cost:", cost2)

print()
print("The cheapest path is:", best_path, "with cost", best_cost)

# Assuming bfs and dfs functions are defined in previous cells and are accessible
bfs_order = bfs("A", "G")
dfs_order = dfs("A", "G")

print("\nRESULTS")
print("-" * 40)

print("BFS (nearest first) checked:", len(bfs_order), "rooms:", bfs_order)
print("DFS (deep first) checked:", len(dfs_order), "rooms:", dfs_order)
print("Cheapest path:", best_path, "with cost", best_cost)

print("-" * 40)
```
